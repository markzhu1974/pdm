# PCA（主成分分析）详细数学推导

## 一、PCA的本质目的

主成分分析（PCA）的根本目的是：

> 在降低数据维度的同时，尽可能保留原数据的“变化信息”（即方差信息）。

换句话说，我们希望从原始数据的多个变量中，找到一组新变量（叫主成分），这些新变量是原变量的线性组合，并且能解释原始数据中最多的方差信息。

---



## 二、问题定义（明确每个数学对象）

假设原始数据如下：

- 样本数： $n$
- 每个样本维度： $d$
- 数据矩阵： $X \in \mathbb{R}^{n \times d}$

$$
X = \begin{bmatrix}
(x^{(1)})^T \\
(x^{(2)})^T \\
\vdots \\
(x^{(n)})^T
\end{bmatrix}
$$

其中：

- 每个样本 $x^{(i)} \in \mathbb{R}^d$ 是**列向量**，被转置后作为矩阵 $X$ 的行。

---



## 三、数据中心化（非常重要）

PCA前提：数据需中心化，消除样本的平均偏移。

**每个维度的样本均值（Mean vector）**：

$$
\bar{x} = \frac{1}{n} \sum_{i=1}^n x^{(i)} \in \mathbb{R}^d
$$

中心化操作：

$$
\tilde{x}^{(i)} = x^{(i)} - \bar{x} \quad \forall i=1,2,\cdots,n
$$

构成新的中心化矩阵：

$$
\tilde{X} = \begin{bmatrix}
(\tilde{x}^{(1)})^T \\
(\tilde{x}^{(2)})^T \\
\vdots \\
(\tilde{x}^{(n)})^T
\end{bmatrix} = X - \mathbf{1}_n \bar{x}^T
$$

其中：

- $\mathbf{1}_n \in \mathbb{R}^{n \times 1}$：全为1的列向量；
- $\bar{x}^T \in \mathbb{R}^{1 \times d}$：样本均值（行向量）；
- $\mathbf{1}_n \bar{x}^T \in \mathbb{R}^{n \times d}$：每行都是样本均值。

此时 $\tilde{X}$ 每一列的均值为0。

---



## 四、协方差矩阵推导（详细到每一步）

协方差矩阵 $C$ 捕捉各特征之间的关系，定义为：

$$
C = \frac{1}{n} \tilde{X}^T \tilde{X} \in \mathbb{R}^{d \times d}
$$

展开来看：

$$
C_{pq} = \frac{1}{n} \sum_{i=1}^n \tilde{x}^{(i)}_p \tilde{x}^{(i)}_q
$$

含义：

- $C_{pp}$：第 $p$ 个特征的方差；
- $C_{pq}$：第 $p$ 和第 $q$ 个特征的协方差；
- 对称：$C_{pq} = C_{qp}$。

---



## 五、PCA的优化目标

寻找方向 $w \in \mathbb{R}^d$，使得中心化数据投影到该方向后，**投影方差最大**。

投影后数据为：

$$
z^{(i)} = w^T \tilde{x}^{(i)} \quad \forall i
$$

所有样本的投影方差：

$$
\text{Var}(z) = \frac{1}{n} \sum_{i=1}^n \left( w^T \tilde{x}^{(i)} \right)^2 = \frac{1}{n} \sum_{i=1}^n w^T \tilde{x}^{(i)} (\tilde{x}^{(i)})^T w
$$

组合为矩阵形式：

$$
\text{Var}(z) = w^T \left( \frac{1}{n} \sum_{i=1}^n \tilde{x}^{(i)} (\tilde{x}^{(i)})^T \right) w = w^T C w
$$

所以最大化方差问题变为：

$$
\max_{w} \quad w^T C w \quad \text{s.t.} \quad w^T w = 1
$$

这里的约束是避免 $w$ 无限变大（投影无限大）——我们限制 $w$ 为单位向量。

---



## 六、拉格朗日乘子法求解（详细展开）

构建Lagrangian：

$$
L(w, \lambda) = w^T C w - \lambda(w^T w - 1)
$$

对 $w$ 求偏导，令导数为0：

$$
\frac{\partial L}{\partial w} = 2Cw - 2\lambda w = 0
$$

解得：

$$
Cw = \lambda w
$$

说明：

- $w$：协方差矩阵的特征向量；
- $\lambda$：对应的特征值。

这表示**主成分方向即为协方差矩阵的特征向量**，且方差大小正比于对应特征值。

---



## 七、选取主成分（排序）

- 按特征值从大到小排序：

$$
\lambda_1 \geq \lambda_2 \geq \cdots \geq \lambda_d
$$

- 选择前 $k$ 个特征向量 $w_1, w_2, \cdots, w_k$ 构成投影矩阵：

$$
W = [w_1, w_2, \dots, w_k] \in \mathbb{R}^{d \times k}
$$

---



## 八、数据降维

将原数据投影到新空间：

$$
Z = \tilde{X} W \in \mathbb{R}^{n \times k}
$$

- $Z$：降维后的数据；
- 每行：一个样本的 $k$ 维表示；
- 这个新空间内，各主成分之间**完全不相关**（协方差为0）。

---



## 九、PCA几何解释

- 协方差矩阵 $C$：数据分布的“形状”；
- 特征向量：该形状的“主轴”方向；
- 特征值：主轴的“长度”（方差信息大小）；
- 选择最大的几个特征值对应方向投影，保留最大的信息量。

---



## 十、完整总结（避免知识跳跃）

| 步骤 | 数学过程 | 说明 |
|------|---------|------|
| 1. 中心化 | $\tilde{X} = X - \mathbf{1}_n \bar{x}^T$ | 均值为0 |
| 2. 协方差矩阵 | $C = \frac{1}{n} \tilde{X}^T \tilde{X}$ | 特征间关系 |
| 3. 特征值问题 | $Cw = \lambda w$ | 最大方差方向 |
| 4. 投影矩阵 | $W = [w_1, ..., w_k]$ | 选前k个方向 |
| 5. 新坐标系 | $Z = \tilde{X}W$ | 数据降维 |

---



## 十一、推导终极图解（脑中想象）

1. **原坐标系** → 数据呈椭球形分布。
2. 计算协方差矩阵 → 得到这个椭球的“主轴”方向。
3. 特征值大 → 该方向信息量大。
4. 选最大k个方向 → 投影 → 得到“信息最多”的低维数据。

---



## 十二、核心公式总结（仅公式）

$$
C = \frac{1}{n} \tilde{X}^T \tilde{X}
$$

$$
Cw = \lambda w
$$

$$
Z = \tilde{X} W
$$
