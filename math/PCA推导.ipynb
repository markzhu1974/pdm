## ​​基于协方差矩阵的PCA完整数学推导（超详细版）​

### **1. 数据预处理：中心化（Mean Centering）的严格推导**

设原始数据矩阵 $ \mathbf{X} \in \mathbb{R}^{n \times d} $，其元素 $ x_{ki} $ 表示第 $ k $ 个样本的第 $ i $ 个特征值。


设原始数据矩阵 $X \in \mathbb{R}^{n \times d}$ 的完整形式为：

$$
X = \begin{bmatrix}
x_{11} & x_{12} & \cdots & x_{1d} \\
x_{21} & x_{22} & \cdots & x_{2d} \\
\vdots & \vdots & \ddots & \vdots \\
x_{n1} & x_{n2} & \cdots & x_{nd}
\end{bmatrix}
$$

维度说明
- **行方向（$n$ 行）**：每个样本 $\mathbf{x}_i = [x_{i1}, x_{i2}, \dots, x_{id}]^\top$
- **列方向（$d$ 列）**：每个特征 $\mathbf{f}_j = [x_{1j}, x_{2j}, \dots, x_{nj}]^\top$



#### **步骤1.1 计算特征均值**

对每个特征 $ i $（$ i = 1, \dots, d $），计算均值：

$$
\bar{x}_i = \frac{1}{n} \sum_{k=1}^n x_{ki}
$$

均值向量：

$$
\bar{\mathbf{x}} = [\bar{x}_1, \dots, \bar{x}_d]^\top \in \mathbb{R}^d
$$

#### **步骤1.2 中心化变换**

定义中心化数据矩阵 $ \mathbf{X}_{\text{centered}} $，其元素为：

$$
(\mathbf{X}_{\text{centered}})_{ki} = x_{ki} - \bar{x}_i
$$

矩阵形式：

$$
\mathbf{X}_{\text{centered}} = \mathbf{X} - \mathbf{1}_n \bar{\mathbf{x}}^\top
$$

其中 $ \mathbf{1}_n \in \mathbb{R}^n $ 为全1列向量。

#### **验证中心化性质**：

$$
\frac{1}{n} \sum_{k=1}^n (\mathbf{X}_{\text{centered}})_{ki} = \frac{1}{n} \sum_{k=1}^n (x_{ki} - \bar{x}_i) = \bar{x}_i - \bar{x}_i = 0
$$

##### 符号定义
- **n**：样本数量。
- **$x_{ki}$**：第k个样本的第i个特征值。
- **$\bar{x}_i$**：第i个特征的均值，计算为$\frac{1}{n}\sum_{k = 1}^{n}x_{ki}$。
- **$(X_{centered})_{ki}$**：中心化后的数据矩阵的第k行第i列元素。

##### 逐步推导
1. **中心化数据的定义**：
   $$(X_{centered})_{ki}=x_{ki}-\bar{x}_i$$

2. **计算中心化数据的均值**：
   对中心化后的第i个特征求均值：
   $$\frac{1}{n}\sum_{k = 1}^{n}(X_{centered})_{ki}=\frac{1}{n}\sum_{k = 1}^{n}(x_{ki}-\bar{x}_i)$$

3. **拆分求和项**：
   展开为：
   $$\frac{1}{n}\sum_{k = 1}^{n}x_{ki}-\frac{1}{n}\sum_{k = 1}^{n}\bar{x}_i$$
   - 第一项是$\bar{x}_i$（均值定义）
   - 第二项结果是$\bar{x}_i$（因为$\bar{x}_i$是常数，$\frac{1}{n}\sum_{k = 1}^{n}\bar{x}_i = \frac{1}{n} \cdot n \cdot \bar{x}_i = \bar{x}_i$）

4. **最终结果**：
   $$\bar{x}_i-\bar{x}_i = 0$$

##### 结论
中心化后每个特征的均值严格为零。

##### 中心化数据矩阵 $X_{centered}$ 的展开

中心化过程对每列（特征）减去其均值：

$$
X_{centered} = \begin{bmatrix}
x_{11} - \bar{x}_1 & x_{12} - \bar{x}_2 & \cdots & x_{1d} - \bar{x}_d \\
x_{21} - \bar{x}_1 & x_{22} - \bar{x}_2 & \cdots & x_{2d} - \bar{x}_d \\
\vdots & \vdots & \ddots & \vdots \\
x_{n1} - \bar{x}_1 & x_{n2} - \bar{x}_2 & \cdots & x_{nd} - \bar{x}_d
\end{bmatrix}
$$

其中均值向量 $\bar{\mathbf{x}} = [\bar{x}_1, \bar{x}_2, \dots, \bar{x}_d]^\top$ 的计算：

$$
\bar{x}_j = \frac{1}{n}\sum_{i=1}^{n}x_{ij} \quad \forall j \in \{1,\dots,d\}
$$

##### 转置矩阵 $X_{centered}^\top$ 的展开

转置后矩阵的维度为 $\mathbb{R}^{d \times n}$：

$$
X_{centered}^\top = \begin{bmatrix}
x_{11} - \bar{x}_1 & x_{21} - \bar{x}_1 & \cdots & x_{n1} - \bar{x}_1 \\
x_{12} - \bar{x}_2 & x_{22} - \bar{x}_2 & \cdots & x_{n2} - \bar{x}_2 \\
\vdots & \vdots & \ddots & \vdots \\
x_{1d} - \bar{x}_d & x_{2d} - \bar{x}_d & \cdots & x_{nd} - \bar{x}_d
\end{bmatrix}

### **2. 协方差矩阵的严格推导**

#### **步骤2.1 协方差的定义**

协方差 $C_{ij}$ 衡量特征 $i$ 和 $j$ 的线性相关性：

$$
C_{ij} = \frac{1}{n}\sum_{k=1}^{n}(x_{ki}-\bar{x}_i)(x_{kj}-\bar{x}_j)
$$

这是对中心化数据的乘积求平均。

#### **步骤2.2 矩阵形式的证明**

将 $C_{ij}$ 表示为矩阵乘法：

$$
C_{ij} = \frac{1}{n}\sum_{k=1}^{n}(X_{centered})_{ki}(X_{centered})_{kj}
$$

这正是矩阵 $X_{centered}^{\top}X_{centered}$ 的第 $(i,j)$ 元素。

因此协方差矩阵：

$$
C = \frac{1}{n}X_{centered}^{\top}X_{centered}
$$

#### **维度验证**

- $X_{centered} \in \mathbb{R}^{n \times d}$ ⇒ $X_{centered}^{\top} \in \mathbb{R}^{d \times n}$
- $X_{centered}^{\top}X_{centered} \in \mathbb{R}^{d \times d}$，与 $C$ 维度一致


##### 协方差矩阵的完整形式

协方差矩阵 $C$ 的完整矩阵形式为：

$$
C = \begin{bmatrix}
\frac{1}{n}\sum_{k=1}^n (x_{k1}-\bar{x}_1)^2 & \frac{1}{n}\sum_{k=1}^n (x_{k1}-\bar{x}_1)(x_{k2}-\bar{x}_2) & \cdots & \frac{1}{n}\sum_{k=1}^n (x_{k1}-\bar{x}_1)(x_{kd}-\bar{x}_d) \\
\frac{1}{n}\sum_{k=1}^n (x_{k2}-\bar{x}_2)(x_{k1}-\bar{x}_1) & \frac{1}{n}\sum_{k=1}^n (x_{k2}-\bar{x}_2)^2 & \cdots & \frac{1}{n}\sum_{k=1}^n (x_{k2}-\bar{x}_2)(x_{kd}-\bar{x}_d) \\
\vdots & \vdots & \ddots & \vdots \\
\frac{1}{n}\sum_{k=1}^n (x_{kd}-\bar{x}_d)(x_{k1}-\bar{x}_1) & \frac{1}{n}\sum_{k=1}^n (x_{kd}-\bar{x}_d)(x_{k2}-\bar{x}_2) & \cdots & \frac{1}{n}\sum_{k=1}^n (x_{kd}-\bar{x}_d)^2
\end{bmatrix}
$$

##### 维度验证

1. **$X_{centered} \in \mathbb{R}^{n \times d}$**  
   - 样本数 $n$，特征数 $d$

2. **$X_{centered}^\top \in \mathbb{R}^{d \times n}$**  
   - 转置后特征数 $d$ 为行，样本数 $n$ 为列

3. **$X_{centered}^\top X_{centered} \in \mathbb{R}^{d \times d}$**  
   - 矩阵乘法规则：$(d \times n) \times (n \times d) = (d \times d)$

4. **$C \in \mathbb{R}^{d \times d}$**  
   - 标量 $\frac{1}{n}$ 不改变维度

### **3. 特征分解的详细数学过程**

#### **步骤3.1 协方差矩阵的性质**

$C$ 是对称半正定矩阵：
- **对称性**：$C_{ij} = C_{ji}$
- **半正定性**：对任意 $\mathbf{v} \in \mathbb{R}^d$，有 $\mathbf{v}^\top C \mathbf{v} \geq 0$

#### **步骤3.2 特征分解的存在性**

由谱定理，对称矩阵可对角化：

$$
C = W \Lambda W^\top
$$

其中：
- $W = [\mathbf{w}_1, \dots, \mathbf{w}_d]$ 是正交矩阵（$W^\top W = I$）
- $\Lambda = \text{diag}(\lambda_1, \dots, \lambda_d)$，且 $\lambda_1 \geq \cdots \geq \lambda_d \geq 0$


**谱定理说明（Spectral Theorem）**

**定理表述**：  
对于任意实对称矩阵 $\mathbf{C} \in \mathbb{R}^{d \times d}$，存在：
- 正交矩阵 $\mathbf{W}$（满足 $\mathbf{W}^\top \mathbf{W} = \mathbf{I}_d$）
- 对角矩阵 $\mathbf{\Lambda} = \text{diag}(\lambda_1, \lambda_2, \dots, \lambda_d)$

使得矩阵 $\mathbf{C}$ 可对角化为：
$$
\mathbf{C} = \mathbf{W} \mathbf{\Lambda} \mathbf{W}^\top
$$

**关键说明**：
1. $\mathbf{W}$ 的列向量是 $\mathbf{C}$ 的标准正交特征向量
2. $\mathbf{\Lambda}$ 的对角元素是 $\mathbf{C}$ 的特征值（实数）
3. 该分解称为**谱分解**或**特征分解**


#### **步骤3.3 特征向量的物理意义**

特征向量 $\mathbf{w}_i$ 是数据方差最大的方向：
- **投影方差**：$\text{Var}(X\mathbf{w}_i) = \mathbf{w}_i^\top C \mathbf{w}_i = \lambda_i$



### **4. 主成分选择的最优化推导**

#### 步骤4.1 优化问题形式化
**目标**：找到单位向量 $\mathbf{w}$ 最大化投影方差：
$$
\max_{\mathbf{w}} \mathbf{w}^{\top}\mathbf{C}\mathbf{w} \quad \text{s.t.} \quad \mathbf{w}^{\top}\mathbf{w} = 1
$$

#### 步骤4.2 拉格朗日乘子法
**拉格朗日函数**：
$$
\mathcal{L}(\mathbf{w}, \lambda) = \mathbf{w}^{\top}\mathbf{C}\mathbf{w} - \lambda(\mathbf{w}^{\top}\mathbf{w} - 1)
$$

**求导并令梯度为零**：
$$
\nabla_{\mathbf{w}} \mathcal{L} = 2\mathbf{C}\mathbf{w} - 2\lambda\mathbf{w} = 0 \implies \mathbf{C}\mathbf{w} = \lambda\mathbf{w}
$$

**结论**：
- 最优解 $\mathbf{w}$ 是 $\mathbf{C}$ 的特征向量
- $\lambda$ 为对应的特征值

#### 步骤4.3 主成分的序
**主成分排序规则**：
1. 最大特征值 $\lambda_1$ 对应**第一主成分** $\mathbf{w}_1$（方差最大方向）
2. 次大特征值 $\lambda_2$ 对应**第二主成分** $\mathbf{w}_2$（与 $\mathbf{w}_1$ 正交）
3. 依此类推，$\lambda_1 \geq \lambda_2 \geq \cdots \geq \lambda_d \geq 0$

**性质**：
- 特征向量彼此正交：$\mathbf{w}_i^{\top}\mathbf{w}_j = 0 \ (i \neq j)$
- 投影方差等于特征值：$\text{Var}(\mathbf{X}\mathbf{w}_i) = \lambda_i$
